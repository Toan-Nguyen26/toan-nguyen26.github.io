## B. Notebook khác Script thế nào?

| | Notebook (`.ipynb`) | Script (`.py`) |
|---|---|---|
| Cách chạy | chạy **từng ô**, giữ lại biến trong bộ nhớ | chạy **cả file** từ trên xuống |
| Hợp để | học, thử nghiệm, xem ảnh/biểu đồ ngay | phần mềm chạy thật trên máy trạm, chạy tự động |
| Nhược điểm | chạy lộn xộn thứ tự ô → kết quả sai | không xem được kết quả giữa chừng |

Quy ước trong khóa này: **học bằng notebook, sản phẩm cuối giao bằng script `.py`**.

Chạy ô dưới đây. Nếu có dòng nào báo THIẾU, cài thư viện đó rồi chạy lại ô này.

## E. Biến và kiểu dữ liệu

Biến = một cái tên gắn vào một giá trị trong bộ nhớ.

````python
ma_san_pham = "SP002"      # str   - chuỗi
kich_thuoc  = 50.27        # float - số thực
so_luong    = 120          # int   - số nguyên
dat_chuan   = True         # bool  - đúng/sai

for ten, gia_tri in [("ma_san_pham", ma_san_pham), ("kich_thuoc", kich_thuoc),
                     ("so_luong", so_luong), ("dat_chuan", dat_chuan)]:
    print(f"{ten:12} = {str(gia_tri):8} -> kiểu {type(gia_tri).__name__}")
````

````python
# Dữ liệu đọc từ file, từ máy đo, từ ô nhập liệu LUÔN là chuỗi -> phải ép kiểu
chuoi_do = "50.27"
print(chuoi_do + "1")           # nối chuỗi, không phải cộng số!
print(float(chuoi_do) + 1)      # ép sang số rồi mới cộng

print(int("120"), float("0.25"), str(50.27))
````

**Quy tắc đặt tên biến:** chữ thường, nối bằng `_`, có nghĩa.
`kich_thuoc_do_duoc` tốt hơn `x`, `kt`, `a1` — vì 6 tháng sau bạn (hoặc đồng nghiệp trực ca đêm) vẫn phải đọc được.

## F. In báo cáo đẹp bằng f-string

Đây là thứ dùng hằng ngày khi in kết quả kiểm tra.

````python
ma = "SP002"
do_duoc = 50.2734
danh_nghia = 50.0
sai_lech = do_duoc - danh_nghia

print(f"Mã sản phẩm: {ma}")
print(f"Sai lệch: {sai_lech:.2f} mm")        # .2f = làm tròn 2 chữ số thập phân
print(f"Sai lệch: {sai_lech:+.3f} mm")       # +  = luôn hiện dấu
print(f"Tỉ lệ NG: {3/12:.1%}")               # .1% = phần trăm
print(f"|{ma:<10}|{do_duoc:>10.2f}|")        # căn trái / căn phải -> làm bảng
````

## G. Nhận dữ liệu từ người vận hành: `input()`

````python
try:
    ma_nhap = input("Nhập mã sản phẩm: ")
except Exception:          # phòng khi notebook chạy tự động, không có người gõ
    ma_nhap = ""

ma_nhap = ma_nhap.strip().upper() or "SP002"    # bỏ khoảng trắng thừa, viết hoa, rỗng thì lấy mặc định
print("Đang kiểm tra:", ma_nhap)
````

> `input()` **luôn trả về chuỗi**. Muốn tính toán thì phải `float(...)`.
> `.strip()` và `.upper()` là hai hàm phải có khi nhận dữ liệu tay — người vận hành rất hay gõ thừa dấu cách hoặc viết thường.

## H. Ba lỗi bạn chắc chắn sẽ gặp hôm nay

| Thông báo | Nghĩa là | Cách sửa |
|---|---|---|
| `IndentationError` | thụt lề sai | khối lệnh trong `if`/`for` phải thụt vào **4 dấu cách**, thống nhất cả file |
| `NameError: name 'x' is not defined` | dùng biến chưa tạo (hoặc chưa chạy ô phía trên) | chạy lại các ô từ trên xuống, kiểm tra chính tả tên biến |
| `TypeError: can only concatenate str...` | cộng chuỗi với số | ép kiểu: `float(chuoi)` hoặc `str(so)` |

Mẹo đọc lỗi: **đọc dòng cuối cùng trước**, rồi nhìn số dòng ở ngay trên nó.

````python
# Thử cho lỗi xảy ra để quen mặt (ô này cố tình sai, chạy xong nhớ đọc dòng cuối)
try:
    print("Kích thước: " + 50.27)
except TypeError as loi:
    print("TypeError ->", loi)
    print("Sửa:", "Kích thước: " + str(50.27))
````

### Gợi ý lời giải

```python
DANH_NGHIA, DUNG_SAI, GIOI_HAN_LOI = 50.0, 0.25, 3.0

ma = input("Nhập mã sản phẩm: ").strip().upper()
do_duoc = float(input("Nhập kích thước đo được: "))
dien_tich_loi = float(input("Nhập diện tích khuyết tật: "))

dat_kich_thuoc = abs(sai_lech) <= DUNG_SAI

print(f"\nMã sản phẩm: {ma}")
print(f"Sai lệch kích thước: {sai_lech:.2f} mm")
print(f"Kiểm tra kích thước: {'PASS' if dat_kich_thuoc else 'FAIL'}")
print(f"Kiểm tra bề mặt: {'PASS' if dat_be_mat else 'FAIL'}")

if dat_kich_thuoc and dat_be_mat:
    print("Kết quả cuối cùng: OK")
else:
    print("Kết quả cuối cùng: NG")
    if not dat_kich_thuoc:
        print("Nguyên nhân: Kích thước vượt dung sai")
    if not dat_be_mat:
        print("Nguyên nhân: Khuyết tật bề mặt vượt giới hạn")
```
